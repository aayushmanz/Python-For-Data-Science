```python


lst = [34, 67,45,89,23,90,30, 46]

ans = [lst[i] for i in range(len(lst)) if lst[i] % 2 == 0] # expression iterations condition 
print(ans)
```

```python
ans = [f"{lst[i]}:even" if lst[i] % 2 == 0 else f"{lst[i]}:odd" for i in range(len(lst)) ]
print(ans)
```
