```python
# 1. None Type : 
def welcome (name=None):
    if name is None or name is "":
        print(f"hello , Guest")
    else:
        print(f"hello , {name}")
    
    
if __name__ == "__main__":
    name = input("Enter Your Name : ")
    welcome(name) 
```

```python
# 2. Numeric Type :

def numtype(value):
    match value :
        case 1:
            val = map(int,input().split())
            integer(*val)
            
        case 2:
            val = map(float, input().split())
            floated(*val)
            
        case _:
            print("Invalid Choice !")


def  integer(*args):
    
    value = 0
    
    for i in args:
        value += i
        
    print(f"Your added values are : {value}")    


def floated(*args):
    
    value = 0
    
    for i in args:
        value += i
        
    print(f"the sum of float values : {value}")           
        
        
if __name__ == "__main__":
    a = int(input("Enter the choice 1 for int and 2 for float : "))
    numtype(a)
```

```python
# 4. byte :

element  = [23, 45, 67, 12, 80]

a = bytes(element)

print(a)
print(a[1])
# insert, remove, extend, pop can not be use in byte is mutable


# 5. ByteArray :

element = [24, 56,34, 70]
b = bytearray(element)
b.insert(1, 222)
b[2]=250
print(b[1])
print(b[2])
# insert , remove, extend, pop can be use in bytearray is immutable
```

```python
# 6. tuples :

t1 = (26, 70, 46)
t2 = (45, 68)

res = t1 + t2  # merge
print(res)


v1 = res[:2]
v2 = res[2:5]

print(v1)
print(v2)

v3 = 500
res = v1 + (v3,) + v2

print(res)


tu = (35, 67, 90, 50, 69, 10)  # 0  1  2   -> 3 it will cause us indexError

try :
 a = int(input("Enter the index "))
 if a > 5:
     print(tu[-1])
 else:
     print(tu[a])    
except ValueError:
    print("invalid type !")    
    
    
# tu[2]=56   # tuples are immutable
val1 = tu[0:3]
val2 = tu[4:5+1]

tu = val1 + (5000,) + val2   # updation !



print(tu)
    
 

```

```python
# 7. sets :


sets = {23, 56, 45, 80, "ayush", "suthar", 12.80}

print(56 in sets) # membership function ( in and not in )

value = input("Enter the value already exists : ")
revalue = input("Enter replace value : ")

if value in sets :
    sets.remove(value)
    sets.add(revalue)
    sets.pop()
else :
    print('Invalid !')    
    
print(sets)
```
