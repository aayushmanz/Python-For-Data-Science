## 2. Matrix Plot:

* Heatmap
* clustermap

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import plotly.express as px 

gap = px.data.gapminder()
```

```python
# heatmap
# plot rectangular data as a color - encoder matrix
temp = gap.pivot(index='country', columns='year', values='lifeExp')

# axes level function
plt.figure(figsize=(15,15))
sns.heatmap(temp)
```

```python
# annot
euro = gap[gap['continent']=='Europe'].pivot(index='country', columns='year', values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(euro, annot=True)
```

```python
# linewidth

euro = gap[gap['continent']=='Europe'].pivot(index='country', columns='year', values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(euro, annot=True, linewidths=1.2)
```

```python
# cmap

euro = gap[gap['continent']=='Europe'].pivot(index='country', columns='year', values='lifeExp')
plt.figure(figsize=(15,15))
sns.heatmap(euro, annot=True, linewidths=0.5, cmap='viridis')
```

```python
# clustermap

# Plot a matrix dataset as a hierarchically-clustered heatmap.
# This function requires scipy to be available.

iris = px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,[0,1,2,3]])
```
