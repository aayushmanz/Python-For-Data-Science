```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import plotly.express as px 

tips = sns.load_dataset('tips')
tips
```

```python
# scatter  plot :

# scatter plot is a axis level function.

sns.scatterplot(data=tips, x='total_bill', y='tip')
```

```python
# this is figure level function :

sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter')
```

```python
# scatterplot -> size and hue
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
# scatter  using  relplot -> size and hue
sns.relplot(data=tips, x='total_bill', y='tip', hue='sex', style='time', size='size')
```

```python
# line plot
gap = px.data.gapminder()

india = gap[gap['country']=='India']
india
```

```python
# axes level function :
sns.lineplot(data=india, x='year', y='lifeExp')
```

```python
# using relpplot

sns.relplot(data=india, x='year', y='lifeExp', kind='line')
```

```python
# hue -> style

temp = gap[gap['country'].isin(['India', 'Brazil', 'France'])]
temp
```

```python
sns.relplot(data=temp,x='year', y = 'lifeExp', hue='country', kind='line' )
```

```python
sns.lineplot(data=temp,x='year', y = 'lifeExp', hue='country' )
```

```python

sns.relplot(data=temp,x='year', y = 'lifeExp', hue='country', kind='line', style='continent', size='continent' )
```
