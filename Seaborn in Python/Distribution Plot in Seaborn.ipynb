```python
# Imports  : 

import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import plotly.express as px 

tips = sns.load_dataset('tips')
tips
```

## 2. Distribution Plots

* used for univariate analysis
* used to find out the distribution
* Range of the observation
* Central Tendency
* is the data bimodal?
* Are there outliers?

Plots under distribution plot

* histplot
* kdeplot
* rugplot


```python
# figure level function -> displot
# axes level -> hisplot -> kdeplot -> rugplot
```

```python
# plotting univariate histogram :

sns.histplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='hist')
```

```python
# bins parameter 
sns.displot(data=tips, x='total_bill', kind='hist', bins=20)
```

```python
sns.displot(data=tips, x='total_bill', kind='hist', bins=2)
```

```python
# It’s also possible to visualize the distribution of a categorical variable
# using the logic of a histogram.
# Discrete bins are automatically set for categorical variables.
tips 

# countplot 
sns.displot(data=tips, x='day', kind='hist')
```

```python
# hue parameter :
sns.displot(data=tips, x='tip', kind='hist', hue='sex')
```

```python
# element -> steps 
sns.displot(data=tips, x='tip', kind='hist', hue='sex', element='step')
```

```python
titanic = sns.load_dataset('titanic')
titanic
```

```python
sns.displot(data=titanic, x='age', kind='hist', element='step', hue="sex")
```

```python
# faceting using cols and rows -> this function will not work on histplot function  
sns.displot(data=tips, x='tip', kind='hist', col='sex', element='step', hue='sex')
```

```python
# kdeplot :
# Rather than using discrete bins, a KDE plot smooths the observations
# with a Gaussian kernel, producing a continuous density estimate
sns.kdeplot(data=tips, x='total_bill')
```

```python
sns.displot(data=tips, x='total_bill', kind='kde')
```

```python
# hue => fill
sns.kdeplot(data=tips, x='total_bill', hue='sex', fill=True)
```
