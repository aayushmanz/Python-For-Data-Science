```python
import pandas as pd 
import numpy as np
```

```python
df = pd.read_csv('/workspaces/Python-For-Data-Science/Database from Kaggle/Dirty_data.csv')

df
```

```python
df.isnull().sum()
```

```python
# ok so both columns like peak and all time have more than  70 percent of them are null.
# so decide to remove them from here.

df = df.drop(columns=['Peak', 'All Time Peak'])
df
```

```python
# now i m gonna declare the rank as a set_index() ( as a index )

df = df.set_index('Rank')
df
```

```python
df = df.rename(columns={'Actual\xa0gross' : 'Actual gross'})

df['Actual gross'] = df['Actual gross'].str.split('[').str[0]

df
```

```python
# ok finally this one is also cleaned !

df['Tour title'] = df['Tour title'].str.split('[').str[0]
df['Tour title'] = df['Tour title'].str.split('*').str[0]
df['Tour title'] = df['Tour title'].str.split('†').str[0]
df['Tour title'] = df['Tour title'].str.split('‡').str[0]
df['Tour title'] = df['Tour title'].str.split('2016').str[0]


df
```

```python
# deleting the ref. cols
df.drop(columns='Ref.', inplace=True)

df
```

```python
# ok so that's clear that all the values of the adjusted gross is greater and sometimes equals to actual.

df = df.rename(columns={
    'Adjusted\xa0gross (in 2022 dollars)' : 'Adjusted gross'
})

df['Actual gross']<=df['Adjusted gross']
```

```python
# only in this tour we have equal for adjust and actual gross.

df[df['Adjusted gross']==df['Actual gross']]
```

```python
# these are our top 3 

df.groupby(['Artist','Tour title'])['Adjusted gross'].max().sort_values(ascending=False).head(3)

    # the max gross is taylor swift.
```

```python
# cher have a highest shows.

df.groupby(['Artist','Tour title'])['Shows'].max().sort_values(ascending=False).head()
```

```python
df['Average gross'] = df['Average gross'].str.replace('$', '', regex=False)
df['Average gross'] = df['Average gross'].str.replace(',', '', regex=False)
df['Average gross'] = df['Average gross'].astype(float)
df.groupby('Year(s)')['Average gross'].max().sort_values(ascending=False)
```
