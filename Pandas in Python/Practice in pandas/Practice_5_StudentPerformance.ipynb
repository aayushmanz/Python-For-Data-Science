```python
import pandas as pd

std = pd.read_csv('/content/StudentsPerformance.csv')
std.head(30)
```

```python
std.info()
```

```python
std = std.rename(columns={
    "race/ethnicity":"race",
    "parental level of education" : "edu_level",
    "test preparation course" : "test_prep",
    "math score" :"math",
    "reading score" : "reading",
    "writing score" : "writing"

})


std
```

```python
def average_score(std):
  add = (std['writing'] + std['math'] + std['reading'])
  ans = add / 3
  return ans
```

```python
std['average_score'] = std.apply(average_score, axis=1)
std['average_score']  = pd.to_numeric(std['average_score'])
std['average_score'] = std['average_score'].round(2)
std
```

```python
# std['average_score'] = (std['writing'] + std['math'] + std['reading']) / 3
std
```

```python
none_std = std[std['test_prep']=="none"]
none_std.shape[0]
# ok so total number of none is 642
```

```python
std.head()
```

```python
none_std.groupby('race')['test_prep'].count().sort_values(ascending=False)

# so we find out that the group c have the most number of none prepare student.
```

```python
none_std.groupby(['race','test_prep'])['average_score'].mean().sort_values(ascending=False)

# so the group E students who never prepare for the exam but give the highest number of score.
```

```python
def marks(score):
  if score >= 33:
    return "pass"
  else:
    return "fail"
```

```python
std['final_result'] = std['average_score'].apply(marks)
std[std['final_result']=='pass'].groupby('gender').size()
```
