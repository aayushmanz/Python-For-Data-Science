```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/workspaces/Python-For-Data-Science/Database from Kaggle/Messy_Employee_dataset.csv')

df.sample(20)
```

```python
# insert two columns and dropped the older one.

df.insert(1, 'Name', df['First_Name'] + ' ' + df['Last_Name'])
df.drop(columns=['First_Name', 'Last_Name'], inplace=True)
df
```

```python
df['Age'].isnull().value_counts()  # wow 200+ values are null.
```

```python
# Age :

count_age = df['Age'].value_counts()
plt.pie(count_age, labels=count_age.index, autopct="%1.1f%%")
plt.show()

```

```python
Age_num = np.random.choice([30.0, 35.0, 25.5, 40.0])
df.replace(np.nan , Age_num, inplace=True)

df.sample(10)
```

```python
age_count = df['Age'].value_counts()

plt.pie(age_count, labels=age_count.index, autopct="%1.1f%%")
plt.show()
```

```python
df.sample(30)
```

```python
# salary :

# in salary we can see that some people salary is only 30 dollars
# thats why i am converting them into 30000 dollars.


df['Salary']= df['Salary'].replace(30.00, 30000.00)
df
```

```python
# Email :

df['Email'] = df['Email'].str.replace('example', 'gmail')
df
```

```python
# task 2 :

# department :

df.insert(3,'Department',df['Department_Region'].str.split('-').str[0])
df.insert(4,'Region',df['Department_Region'].str.split('-').str[1])
df.drop(columns='Department_Region', inplace=True)
```

```python
df.groupby('Department')['Salary'].mean().sort_values(ascending=False).head(1)
```

```python
# task 3 : 

join_date = pd.to_datetime(df['Join_Date'])
date_today = pd.to_datetime('today')   # i am not good in date_time so i use my repo and google for the syntax

df.insert(5,'Experience_Years',round((date_today -  join_date).dt.days/365))
df['Experience_Years'] = pd.to_numeric(df['Experience_Years'], errors='coerce')

df
```

```python
df.groupby('Performance_Score')['Experience_Years'].mean().sort_values(ascending=False)
```

```python
df.pivot_table(columns='Performance_Score', values="Experience_Years", aggfunc='mean')
```

```python
# task 4 : 

inactive = df['Status']=='Inactive'

df.pivot_table(index=inactive, columns='Performance_Score', values='Salary')

# ok so in first reason the poor performers are demanding the high salary.
```
