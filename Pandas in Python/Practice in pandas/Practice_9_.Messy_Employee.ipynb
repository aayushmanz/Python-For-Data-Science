```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('/workspaces/Python-For-Data-Science/Database from Kaggle/Messy_Employee_dataset.csv')

df.sample(20)
```

```python
df.info()
```

```python
df.head()
```

```python
# insert two columns and dropped the older one.

df.insert(1, 'Name', df['First_Name'] + ' ' + df['Last_Name'])
df.drop(columns=['First_Name', 'Last_Name'], inplace=True)
df
```

```python
df['Age'].isnull().value_counts()  # wow 200+ values are null.
```

```python
# Age :

count_age = df['Age'].value_counts()
plt.pie(count_age, labels=count_age.index, autopct="%1.1f%%")
plt.show()

```

```python
Age_num = np.random.choice([30.0, 35.0, 25.5, 40.0])
df.replace(np.nan , Age_num, inplace=True)

df.sample(10)
```

```python
age_count = df['Age'].value_counts()

plt.pie(age_count, labels=age_count.index, autopct="%1.1f%%")
plt.show()
```

```python
df.sample(30)
```

```python
# salary :

# in salary we can see that some people salary is only 30 dollars
# thats why i am converting them into 30000 dollars.


df['Salary']= df['Salary'].replace(30.00, 30000.00)
df
```

```python
# Email :

df['Email'] = df['Email'].str.replace('example', 'gmail')
df
```

```python
# active :

status = df['Status'].value_counts()

plt.pie(status,labels=status.index)


# we can clearly see that the pending is 1/3 of the pie chart 
# and we are not capable to change pending into the inactive 
# becuz if we do that then our majority of pie (2/3) will be inactive.
```
