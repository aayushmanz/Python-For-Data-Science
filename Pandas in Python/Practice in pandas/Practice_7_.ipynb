```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt


df = pd.read_csv('/workspaces/Python-For-Data-Science/Database from Kaggle/netflix_titles.csv')
df.sample(20)
```

```python
df[['director','cast', 'country']].isnull().sum()
```

```python
df['director']= df['director'].replace(np.nan, 'Unknown')
df['cast']= df['cast'].replace(np.nan, 'Unknown')
df['country']= df['country'].replace(np.nan, 'Unknown')

df.dropna(subset='date_added', inplace=True)
df.dropna(subset='duration', inplace=True)
df.dropna(subset='rating', inplace=True)

df.dropna(inplace=True) # for final clean !
df.head(10)
```

```python
df.isnull().sum() # ok so now our data has been cleaned ! 
```

```python
df['type'].value_counts()
```

```python
movies = df[df['type']=='Movie']
movies = movies[movies['director']!='Unknown']
movies['director'].value_counts().head(10).plot(kind='bar')
 # ok so there are the most movies directors.
```
