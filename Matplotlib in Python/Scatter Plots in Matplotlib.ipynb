```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/workspaces/Python-For-Data-Science/Matplotlib in Python/Dataset for matplotlib/batter.csv')
```

```python
df = df.head(50)
df
```

```python
plt.style.use('dark_background')
plt.grid()
plt.scatter(df['avg'], df['strike_rate'], color='yellow', marker='+')
plt.title('Avg and SR Analysis of 50 batsman')
plt.xlabel('Avg')
plt.ylabel('Strike rate')
```
