# Bar Chart

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# simple bar chart 

plt.style.use('dark_background')
children = [10,20,40,30,20]
color = ['red','blue', 'yellow', 'green', 'pink']

plt.bar(color, children, color='white')
```

```python
# bar chart using data
```

```python
# for horizontal bar chart
plt.barh(color, children, color='white')
```

```python
# color and label

df = pd.read_csv('/workspaces/Python-For-Data-Science/Matplotlib in Python/Dataset for matplotlib/batsman_season_record.csv')
df
```

```python
df.plot(x='batsman', y=['2015', '2016', '2017'], kind='bar', color=['yellow', 'orange', 'green'], width=0.8)
plt.show()
```

```python
plt.bar(np.arange(df.shape[0]) - 0.2, df['2015'], width=0.2, color='red')
plt.bar(np.arange(df.shape[0]), df['2016'], width=0.2, color='yellow')
plt.bar(np.arange(df.shape[0]) + 0.2, df['2017'], width=0.2, color='blue')

plt.xticks(np.arange(df.shape[0]), df['batsman'])
plt.show()
```

```python
# a problem

children = [10,20,40,10,30]

colors = [
    'red red red red red',
    'blue blue blue blue',
    'green green green green green',
    'yellow yellow',
    'black black'
]

plt.bar(colors, children, color='white')
plt.xticks(rotation='vertical')
plt.show()
```

```python
# stacked bar chart

plt.bar(df['batsman'], df['2017'], color='red', label='2017')
plt.bar(df['batsman'], df['2016'],bottom=df['2017'], color='pink', label='2016')
plt.bar(df['batsman'], df['2015'],bottom=(df['2016'] + df['2017']), color='purple', label='2015')

plt.legend()
plt.show()
```
